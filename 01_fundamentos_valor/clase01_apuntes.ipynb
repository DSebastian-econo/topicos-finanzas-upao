## 1. Valor intrínseco vs. precio de mercado

El **precio** ($P$) es lo que el mercado cobra hoy: observable. El **valor intrínseco** ($V$) es lo que el activo vale para un inversionista que entiende completamente sus características: no observable, se estima.

Cuando tu estimación ($V_E$) difiere del precio, la brecha tiene dos componentes:

$$\underbrace{V_E - P}_{\text{mispricing percibido}} = \underbrace{(V - P)}_{\text{mispricing verdadero}} + \underbrace{(V_E - V)}_{\text{error de estimación}}$$

**Idea fuerza:** parte de la diferencia con el precio puede ser puro error tuyo. Por eso el analista profesional trabaja con margen de seguridad y análisis de sensibilidad, y por eso una recomendación de compra supone dos cosas: que tu error de estimación es pequeño y que existe un catalizador para que el precio converja al valor.

## 3. Valor del dinero en el tiempo: VP, VPN y TIR

Todo activo financiero es una promesa de flujos futuros. Su valor hoy:

$$VP = \sum_{t=1}^{n} \frac{CF_t}{(1+r)^t} \qquad\qquad VPN = -CF_0 + VP$$

donde $r$ es la tasa que **exigimos** por esperar y por arriesgar. La **regla de decisión**: $VPN > 0$ crea valor (aceptar); $VPN < 0$ destruye valor (rechazar). La **TIR** es la tasa que hace $VPN = 0$.

**Lectura financiera:** comprar una acción por debajo de su valor intrínseco es un proyecto de VPN positivo. Valorar empresas y evaluar proyectos son la misma idea.

```python
import sys; sys.path.append("..")
from utils.finanzas import vp, vpn, tir

# El ejemplo de la clase: invertir 100, recibir 30, 40 y 60, con r = 10%
print(f"VPN al 10% = {vpn(100, [30, 40, 60], 0.10):.2f}   (> 0: se acepta)")
print(f"VPN al 15% = {vpn(100, [30, 40, 60], 0.15):.2f}   (< 0: se rechaza)")
print(f"TIR        = {tir(100, [30, 40, 60]):.2%}  (la tasa que separa ambos mundos)")
```

**Idea fuerza:** el mismo proyecto se acepta al 10% y se rechaza al 15%. La tasa de descuento decide, y por eso la semana 2 está dedicada a estimarla (CAPM y WACC).

## 4. Creación de valor para el accionista

Una empresa crea valor cuando el retorno de lo que invierte supera el costo de su capital:

$$\text{Crea valor} \iff ROIC > WACC \qquad\qquad EVA = (ROIC - WACC) \times \text{Capital invertido}$$

**Trampa clásica:** crecer no es sinónimo de crear valor. Crecer invirtiendo a $ROIC < WACC$ destruye valor más rápido cuanto más se crece.

## 5. La tríada riesgo, retorno y valor

A flujos iguales: más riesgo, mayor tasa exigida, menor valor presente. La pregunta de cuánto más de tasa por cuánto más de riesgo la responde el CAPM (semana 2), y en la Unidad 3 veremos que esa tasa también se mueve con la macroeconomía.
