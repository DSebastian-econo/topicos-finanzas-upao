## 1. El costo de capital y la regla de consistencia

La tasa de descuento es el **costo de oportunidad** de quienes financian a la empresa: accionistas ($K_e$) y acreedores ($K_d$). La empresa como un todo se descuenta a su promedio ponderado, el WACC.

**Regla de consistencia** (cae en todos los exámenes): los flujos para toda la firma (FCFF) se descuentan al **WACC**; los flujos solo para el accionista (FCFE) se descuentan al **costo del equity** $K_e$. Mezclar flujo de uno con tasa del otro es el error clásico.

## 2. El CAPM y sus tres insumos

$$E(R_i) = R_f + \beta_i \left[ E(R_m) - R_f \right]$$

- $R_f$: **tasa libre de riesgo**. Bono soberano de largo plazo (10 años), en la misma moneda que los flujos, nominal con nominal.
- $E(R_m) - R_f$: **prima por riesgo de mercado (ERP)**. No se observa: histórica (4% a 6% en EE.UU.), implícita (Damodaran) o de encuestas. Se declara y se justifica.
- $\beta_i$: cuánto **riesgo de mercado** carga la acción. El mercado solo paga por el riesgo que no se puede diversificar.

## 3. El beta: estimación y apalancamiento

El beta se estima por regresión (5 años de retornos mensuales contra el mercado) y viene con error de estimación: mira siempre su error estándar. El **ajuste de Blume** reconoce que los betas tienden a 1:

$$\beta_{aj} = 0.67\,\beta_{OLS} + 0.33$$

El beta del equity mezcla riesgo de negocio y de apalancamiento. Para transferir betas entre empresas (ignorando el beta de la deuda):

$$\beta_U = \frac{\beta_E}{1 + (1-t)\,D/E} \qquad\qquad \beta_E = \beta_U\left[1 + (1-t)\,D/E\right]$$

```python
import sys; sys.path.append("..")
from utils.finanzas import capm, wacc, beta_ajustado, beta_desapalancar, beta_reapalancar

# Ejercicio de la clase: beta_E = 1.4, D/E = 0.8, t = 30%
bu = beta_desapalancar(1.4, 0.8, 0.30)
print(f"Beta desapalancado (del negocio) = {bu:.3f}")
print(f"Reapalancado a D/E = 0.4         = {beta_reapalancar(bu, 0.4, 0.30):.3f}")
print(f"Beta ajustado de Blume para un OLS de 1.2 = {beta_ajustado(1.2):.3f}")
```

**Idea fuerza:** menos deuda, menos beta, menos retorno exigido. El apalancamiento no cambia el negocio; cambia quién carga el riesgo.

## 4. Mercados emergentes: prima por riesgo país

$$K_e = R_f + \beta \times ERP + \lambda \times CRP$$

El CRP se aproxima con el **spread EMBIG** (cuánto más paga el soberano peruano en dólares sobre el Tesoro americano) y $\lambda$ mide la exposición de la empresa al riesgo país (menor si exporta la mayor parte de sus ingresos).

## 5. Costo de la deuda y WACC

$K_d$ es la tasa a la que la empresa podría endeudarse **hoy** (YTM de sus bonos, tasa por rating, o gasto de intereses entre deuda como aproximación). Como los intereses son deducibles, el costo efectivo es $K_d(1-t)$; en el Perú $t = 29.5\%$.

$$WACC = \frac{E}{E+D}\,K_e + \frac{D}{E+D}\,K_d\,(1-t)$$

con $E$ y $D$ a **valor de mercado** y, en teoría, con la estructura de capital objetivo. El WACC es la valla mínima de retorno: conecta con el $ROIC > WACC$ de la semana 1.

```python
# El ejemplo de la clase: E = 700, D = 300, rf = 4.5%, ERP = 5.5%, beta = 1.2, Kd = 6.5%, t = 29.5%
ke = capm(rf=0.045, beta=1.2, erp=0.055)
w = wacc(E=700, D=300, ke=ke, kd=0.065, t=0.295)
print(f"Ke   = {ke:.2%}")
print(f"WACC = {w:.2%}")
print(f"Control de razonabilidad: {0.065*(1-0.295):.2%} < {w:.2%} < {ke:.2%}")

# Sensibilidad: la ERP es el insumo mas discutible
ke_alt = capm(rf=0.045, beta=1.2, erp=0.065)
print(f"\nSi la ERP fuera 6.5%: Ke = {ke_alt:.2%} y WACC = {wacc(700, 300, ke_alt, 0.065, 0.295):.2%}")
```

## 6. Errores comunes (revisar antes de la PC1)

- Usar pesos contables en lugar de valores de mercado en el WACC.
- Mezclar monedas: $K_e$ en dólares con flujos en soles.
- Olvidar el escudo tributario en $K_d$, o aplicarlo dos veces.
- Descontar FCFE al WACC o FCFF al $K_e$ (violación de la regla de consistencia).
- Usar el WACC de la empresa para un proyecto de riesgo distinto.

El WACC siempre debe quedar entre $K_d(1-t)$ y $K_e$: si no, hay un error de insumos.

## 7. Glosario CFA (español - inglés)

| Español | Inglés (así aparece en el CFA) |
|---|---|
| Costo de capital | Cost of capital |
| Costo del capital propio | Cost of equity |
| Prima por riesgo de mercado | Equity risk premium (ERP) |
| Beta apalancado / desapalancado | Levered / unlevered beta |
| Prima por riesgo país | Country risk premium (CRP) |
| Costo promedio ponderado del capital | Weighted average cost of capital (WACC) |
| Escudo tributario | Tax shield |
| Estructura de capital objetivo | Target capital structure |
