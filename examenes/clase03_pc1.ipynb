## Parte 2: encuentra los errores (6 pts)

Un practicante valoró a la empresa Andina SAC y entregó este trabajo. Sus datos de partida son correctos: EBIT = 50, t = 30 por ciento, depreciación = 10, capex = 15, incremento del capital de trabajo = 5, gasto por intereses = 8, endeudamiento neto = 0, Ke = 12 por ciento, WACC = 9 por ciento.

> **Paso 1.** FCFF = EBIT(1-t) + Dep + Int - Capex - WCInv = 35 + 10 + 8 - 15 - 5 = **33**
>
> **Paso 2.** FCFE = FCFF - Int = 33 - 8 = **25**
>
> **Paso 3.** Como la empresa creció 7 por ciento en los últimos años, uso g = 7 por ciento a perpetuidad. Valor del equity = FCFE por (1+g) entre (WACC - g) = 25 por 1.07 entre 0.02 = **1,337.5**

El trabajo contiene **cuatro errores conceptuales**. Identifícalos (2 pts), explica en una oración por qué cada uno es un error (2 pts), y recalcula el valor del equity corrigiendo todo, con un g perpetuo de 3 por ciento (2 pts). Puedes usar la celda de código.

*Errores y explicaciones:*

1.En el FCFF desde EBIT se sumó el interés.Al calcular el FCFF desde EBIT no se debe sumar el interés, porque el EBIT ya representa el resultado antes de intereses.

2.Para pasar de FCFF a FCFE se restó el interés completo. Para obtener el FCFE debe restarse el interés después de impuestos, no el interés completo

3.Usar \(g=7\%\) a perpetuidad solo porque la empresa creció 7% históricamente, es incorrecto.

4.El FCFE fue descontado al WACC. El FCFE corresponde únicamente al accionista, por lo que debe descontarse al costo del equity \(Ke\), no al WACC. La regla vista en la semana 2 es FCFF → WACC y FCFE → Ke.

````python
# Recalculo corregido (puedes usar utils/finanzas.py si trabajas en tu fork)
EBIT = 50
t = 0.30
dep = 10
capex = 15
wc_inv = 5
interes = 8
endeudamiento_neto = 0
ke = 0.12
g = 0.03

fcff = EBIT * (1 - t) + dep - capex - wc_inv

fcfe = fcff - interes * (1 - t) + endeudamiento_neto

equity = fcfe * (1 + g) / (ke - g)

print("FCFF:", fcff)
print("FCFE:", fcfe)
print("Valor del equity:", equity)
# TU CODIGO AQUI
````

## Parte 3: tu empresa (6 pts)

Con la empresa de tus tareas (usa `yfinance` o los resultados que ya tienes de la tarea 2):

**3.1** Calcula el FCFF del último año disponible por la ruta del EBIT. Muestra los insumos que usaste (EBIT, tasa efectiva, depreciación, capex, capital de trabajo). (3 pts)

````python
import yfinance as yf
import pandas as pd

TICKER = "BVN"
empresa = yf.Ticker(TICKER)

income = empresa.income_stmt
cashflow = empresa.cashflow

# Buscar el último año con TODOS los datos necesarios
fechas_comunes = income.columns.intersection(cashflow.columns)

fecha = None

for f in sorted(fechas_comunes, reverse=True):

    datos_income = [
        income.loc["EBIT", f],
        income.loc["Tax Provision", f],
        income.loc["Pretax Income", f],
        income.loc["Reconciled Depreciation", f]
    ]

    datos_cashflow = [
        cashflow.loc["Capital Expenditure", f],
        cashflow.loc["Change In Working Capital", f]
    ]

    if all(pd.notna(x) for x in datos_income + datos_cashflow):
        fecha = f
        break

if fecha is None:
    raise ValueError("No se encontró un año con todos los datos necesarios.")

# Insumos
EBIT = float(income.loc["EBIT", fecha])

impuestos = float(income.loc["Tax Provision", fecha])
utilidad_antes_impuestos = float(income.loc["Pretax Income", fecha])

t = impuestos / utilidad_antes_impuestos

dep = float(income.loc["Reconciled Depreciation", fecha])

capex_raw = float(cashflow.loc["Capital Expenditure", fecha])
capex = abs(capex_raw)

cambio_wc = float(cashflow.loc["Change In Working Capital", fecha])

# Yahoo lo presenta como efecto sobre la caja
wc_inv = -cambio_wc

# FCFF por ruta EBIT
fcff = EBIT * (1 - t) + dep - capex - wc_inv

# Resultados
print("Año utilizado:", fecha.year)
print(f"EBIT: {EBIT/1e6:,.2f} millones")
print(f"Tasa efectiva: {t:.2%}")
print(f"Depreciación: {dep/1e6:,.2f} millones")
print(f"Capex: {capex/1e6:,.2f} millones")
print(f"Inversión en capital de trabajo: {wc_inv/1e6:,.2f} millones")
print("-" * 45)
print(f"FCFF: {fcff/1e6:,.2f} millones")
````

Para Buenaventura utilicé el año 2023, ya que fue el último periodo con información completa disponible en yfinance. Con un EBIT de 180.54 millones, una tasa efectiva de 52.10 %, depreciación de 2.00 millones, capex de 238.67 millones y una inversión en capital de trabajo de -4.95 millones, obtuve un FCFF de -145.24 millones. El FCFF negativo se explica principalmente por el elevado capex de 238.67 millones, que superó ampliamente la caja operativa generada después de impuestos.

````python
# 3.2 Analizar deuda y FCFE de Buenaventura

balance = empresa.balance_sheet

# Trayectoria de la deuda
if "Total Debt" in balance.index:
    deuda = balance.loc["Total Debt"].dropna().sort_index()

    print("Trayectoria de la deuda:")
    for f, valor in deuda.items():
        print(f"{f.year}: {valor/1e6:,.2f} millones")
else:
    print("No se encontró la fila Total Debt")

# Intereses del año usado
interes = float(income.loc["Interest Expense", fecha])

# Endeudamiento neto
if "Net Issuance Payments Of Debt" in cashflow.index:
    endeudamiento_neto = cashflow.loc[
        "Net Issuance Payments Of Debt", fecha
    ]
else:
    endeudamiento_neto = (
        cashflow.loc["Issuance Of Debt", fecha]
        + cashflow.loc["Repayment Of Debt", fecha]
    )

endeudamiento_neto = float(endeudamiento_neto)

# FCFE
fcfe = fcff - interes * (1 - t) + endeudamiento_neto

print("\nAño analizado:", fecha.year)
print(f"FCFF: {fcff/1e6:,.2f} millones")
print(f"Intereses: {interes/1e6:,.2f} millones")
print(f"Endeudamiento neto: {endeudamiento_neto/1e6:,.2f} millones")
print(f"FCFE: {fcfe/1e6:,.2f} millones")
````
