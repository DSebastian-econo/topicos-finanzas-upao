## 1. Valoración relativa: lógica y peligro

La ley de un solo precio: activos equivalentes deberían costar lo mismo. Los múltiplos comparan el precio de tu empresa contra el de sus similares. Es el método más usado en la práctica por rápido y comunicable, y su talón de Aquiles es estructural: si todo el sector está caro, los comparables normalizan lo caro. Mide precio relativo, no valor absoluto. Por eso el profesional triangula: DCF como ancla intrínseca, múltiplos como contraste de mercado.

## 2. P/E y la trampa de consistencia

$$P/E = \frac{P}{EPS}$$

Trailing (últimos 12 meses, observado) contra forward (próximos 12, estimado): nunca se mezclan en un mismo panel. Si las utilidades crecen, el forward es menor que el trailing; comparar tu forward contra el trailing del grupo te hace parecer artificialmente barato.

Problemas del P/E: colapsa con utilidades negativas, es sensible a la contabilidad y mezcla negocio con apalancamiento. Alternativas: P/B (financieras, ligado al ROE) y P/S (sin utilidades, pero ciego a márgenes).

## 3. El múltiplo justificado

$$\left(\frac{P}{E}\right)_{trailing}^{just} = \frac{\text{payout}\,(1+g)}{K_e - g} \qquad\qquad \left(\frac{P}{E}\right)_{forward}^{just} = \frac{\text{payout}}{K_e - g}$$

Conecta la relativa con la intrínseca: qué múltiplo merece la empresa según sus fundamentos. Y la finura de la semana 3: usar la capacidad de pago (FCFE/NI) como payout, no el dividendo, cuando difieren.

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))
from utils.finanzas import per_justificado, valor_por_multiplo_ev

# Nuestra empresa: NI = 56.75, capitalizacion 700 -> P/E de mercado
NI, E = 56.75, 700
print(f"P/E de mercado = {E/NI:.1f}x")

Ke, g = 0.111, 0.04
payout_dividendos = 0.40
payout_capacidad = 46.75 / NI          # FCFE / NI = 82%

print(f"P/E justificado con dividendos (40%)  = {per_justificado(payout_dividendos, Ke, g):.1f}x")
print(f"P/E justificado con capacidad (82%)   = {per_justificado(payout_capacidad, Ke, g):.1f}x")
```

Salida esperada: el mercado paga 12.3x; con el payout de dividendos el justificado da 5.9x (parecería carísima), pero con la capacidad de pago da 12.1x: coherente con el mercado y con nuestro DCF. El dividendo no mide capacidad; el FCFE sí.

## 4. EV/EBITDA y el método de comparables

EV/EBITDA es el estándar para comparar negocios: neutral al apalancamiento (numerador incluye deuda, denominador es pre intereses) e inmune a políticas de depreciación. Sus límites: ignora diferencias de capex y hereda la sobrevaloración sectorial.

El método en 4 pasos: elegir comparables con criterios declarados; elegir el múltiplo; calcular con datos consistentes y resumir con la **mediana** (resiste outliers); aplicar con puente al equity si es múltiplo de firma, más ajustes declarados (tamaño, liquidez, riesgo país).

## 5. El ejemplo integrador y la triangulación

```python
# Comparables mineras: mediana EV/EBITDA de 7.5x. Nuestra empresa: EBITDA 130, deuda 300, 100 acciones
v_comparables = valor_por_multiplo_ev(7.5, 130, 300, 100)
v_per_just = per_justificado(payout_capacidad, Ke, g) * NI / 100

print(f"Por comparables (7.5x EV/EBITDA) = {v_comparables:.2f} por accion")
print(f"Por P/E justificado (12.1x)      = {v_per_just:.2f} por accion")
print()
print("Triangulacion de la unidad 1 (precio de mercado: 7.00):")
for metodo, v in [("DCF dos etapas (Gordon)", 7.69), ("DCF multiplo salida 8x", 7.84),
                  ("Comparables 7.5x", v_comparables), ("P/E justificado", v_per_just)]:
    print(f"  {metodo:26s} {v:5.2f}  ({v/7.00-1:+.1%})")
```

El precio (7.00) cae dentro del rango de los cuatro métodos (6.75 a 7.84): el veredicto profesional es "bien valorada". La dispersión entre métodos es el tamaño de la incertidumbre, no un defecto del análisis.

Un detalle que ordena las ideas: el P/E justificado con payout de capacidad **no es un método relativo**. Es el modelo FCFE de una etapa de la semana 3 escrito como múltiplo: $12.07 \times 56.75 = 684.8$, exactamente el equity que obtuvimos por la ruta FCFE. En la tabla hay entonces tres estimaciones intrínsecas (7.69 y 7.84 del DCF de dos etapas, 6.85 del de una etapa) y una sola relativa (6.75 de comparables). El múltiplo justificado sirve para leer el P/E observado con ojos de fundamentos, no para sumar una opinión de mercado adicional.
