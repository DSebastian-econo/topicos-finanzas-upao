## Parte 2: los cinco errores de Sierra Azul

1. **Tasa libre de riesgo de 3 meses para flujos de largo plazo.** Descalce de horizonte: los flujos que se valoran son de largo plazo, corresponde el Treasury a 10 años (4.5 por ciento). El correcto: Ke = 4.5 + 1.3 x 5 = 11 por ciento.
2. **Pesos del WACC a valor en libros.** Los pesos van a valor de mercado: E = 400 y D = 200, es decir dos tercios y un tercio, no mitad y mitad.
3. **Interés completo en la ruta de la utilidad neta.** Se devuelve Int(1-t) = 7, no 10. Su FCFF quedó inflado en Int x t = 3.
4. **Descontó un FCFF al Ke.** Violación de la regla de consistencia: el FCFF va con el WACC. (Y para llegar al accionista, después hay que restar la deuda: tampoco lo hizo.)
5. **g perpetuo de 8 por ciento.** El crecimiento histórico de corto plazo no es sostenible a perpetuidad; ninguna empresa crece más que la economía para siempre. Además, con g tan cerca de la tasa, el denominador explota y el valor pierde sentido (por eso le salió 1,643.5).

```python
# Recalculo correcto
capm = lambda rf, b, erp: rf + b * erp
wacc_f = lambda E, D, ke, kd, t: E/(E+D)*ke + D/(E+D)*kd*(1-t)

Ke = capm(0.045, 1.3, 0.05)
WACC = wacc_f(400, 200, Ke, 0.06, 0.30)

NI = (60 - 10) * 0.70
FCFF_ni = NI + 15 + 10*0.70 - 20 - 5          # ruta NI con Int(1-t)
FCFF_ebit = 60*0.70 + 15 - 20 - 5             # control por la ruta EBIT

g = 0.03
EV = FCFF_ebit * (1 + g) / (WACC - g)
eq = EV - 200
print(f"Ke = {Ke:.1%} | WACC = {WACC:.2%}")
print(f"FCFF: ruta NI = {FCFF_ni:.0f} | ruta EBIT = {FCFF_ebit:.0f} (coinciden)")
print(f"EV = {EV:,.1f} | equity = {eq:,.1f} | por accion = {eq/100:.2f} vs precio 4.00")
```

Salida esperada: Ke = 11.0 por ciento, WACC = 8.73 por ciento, FCFF = 32 por ambas rutas, EV = 574.9, equity = 374.9 y **3.75 por acción contra un precio de 4.00**.

**Veredicto modelo:** lejos de estar "regaladísima", el modelo corregido sugiere una leve sobrevaloración (el precio está 6 a 7 por ciento por encima del valor estimado), una diferencia dentro del margen de error de estimación: el veredicto honesto es "aproximadamente bien valorada, sujeto a sensibilidad". Nota el contraste: los cinco errores juntos convertían una empresa bien valorada en una "oportunidad" de 4 veces su precio. Así de caro cuesta cada error conceptual.
